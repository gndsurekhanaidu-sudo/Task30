```python

import pandas as pd
import numpy as np

# Creating CSV Dataset
data = {
    'Math': [78, 85, 62, 90, 73, 88, 56, 95, 69, 81],
    'Reading': [82, 88, 65, 92, 75, 90, 60, 96, 72, 84],
    'Writing': [80, 86, 64, 91, 74, 89, 58, 94, 70, 82]
}

df = pd.DataFrame(data)

# Save dataset as CSV file
df.to_csv('StudentsPerformance.csv', index=False)

print("Student Performance Dataset:")
print(df)

# Descriptive Statistics
print("\nDescriptive Statistics:")
print(df.describe())

# Mean, Median and Mode
print("\nMean:")
print(df.mean())

print("\nMedian:")
print(df.median())

print("\nMode:")
print(df.mode().iloc[0])

# Range
print("\nRange:")
print(df.max() - df.min())

# Variance
print("\nVariance:")
print(df.var())

# Standard Deviation
print("\nStandard Deviation:")
print(df.std())

# Quartiles
print("\nQuartiles:")
print(df.quantile([0.25, 0.50, 0.75]))

# Percentiles
print("\nPercentiles:")
print(df.quantile([0.25, 0.50, 0.90]))

print("\nCSV file created successfully!")
```
